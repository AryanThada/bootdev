```python
from urllib.parse import urlsplit

def normalize_url(url: str)-> str:

    url_object = urlsplit(url)
    
    return url_object


o  = normalize_url("scheme://netloc/path?query#fragment")
print(o)

o1 = normalize_url("https://www.boot.dev/")
print(o1)
o2 = normalize_url("https://www.boot.dev:443/blog")
print(o2.netloc)
print(o2.hostname)
print(o2.port)
print(o2.scheme)
print(o2.path)

normalize_string = f"{o.scheme}://{o.netloc}{o.path}?{o.query}#{o.fragment}"

print(normalize_string) 
```

```python
import requests
def get_html(url):

    response_object = requests.get(url , headers = {"User-Agent" : "BootCrawler/1.0"})


    print(response_object.raise_for_status()) 

    print(response_object.headers)


    return response_object.text

html_doc = get_html("https://learnwebscraping.dev/practice/ecommerce/")

html_doc
```
